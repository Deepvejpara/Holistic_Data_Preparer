## Part A: Conceptual Foundation

### 1. Write short notes on:

#### 🔹 What is Data Analysis?
Data Analysis is the process of collecting, cleaning, transforming, and interpreting data to extract useful insights and support decision-making.  
It involves techniques like data visualization, statistical analysis, and exploratory data analysis (EDA) to identify patterns, trends, and relationships within data.

---

#### 🔹 How to Plan a Data Science Project
Planning a Data Science project involves the following steps:

1. **Define the Problem** – Clearly understand the objective or question to solve  
2. **Data Collection** – Gather relevant data from reliable sources  
3. **Data Cleaning** – Handle missing values, outliers, and inconsistencies  
4. **Exploratory Data Analysis (EDA)** – Understand data patterns and relationships  
5. **Feature Engineering** – Create or transform features for better model performance  
6. **Model Selection** – Choose appropriate machine learning algorithms  
7. **Model Training & Evaluation** – Train models and evaluate performance  
8. **Deployment** – Deploy the model for real-world use  
9. **Monitoring & Maintenance** – Continuously monitor and improve the model  

---

#### 🔹 How to Frame a Machine Learning Problem
Framing a Machine Learning problem means converting a real-world problem into a structured ML task:

- **Define Input Features (X)** – Independent variables (e.g., age, BMI, glucose)  
- **Define Target Variable (y)** – Output to predict (e.g., disease risk)  
- **Identify Problem Type**:
  - Classification → Predict categories (e.g., 0 or 1)  
  - Regression → Predict continuous values  
- **Determine Evaluation Metrics**:
  - Accuracy, Precision, Recall (for classification)  
  - RMSE, MAE (for regression)  
- **Understand Constraints** – Data size, computation, business needs  

---

### 2. Explain Tensors with NumPy Examples

#### 🔹 What is a Tensor?
A Tensor is a multi-dimensional array used to represent data in machine learning and deep learning.  
It generalizes:
- Scalar → 0D Tensor  
- Vector → 1D Tensor  
- Matrix → 2D Tensor  
- Higher dimensions → ND Tensor  

---

````python
# Simple Imputer
mean_imputer = SimpleImputer(strategy='mean')

df_copy = df.copy()

df_copy['age'] = mean_imputer.fit_transform(df_copy[['age']]).ravel()
````

````python
most_fq_imputer = SimpleImputer(strategy='most_frequent')

df_copy['employment_type'] = most_fq_imputer.fit_transform(df[['employment_type']]).ravel()
````

````python
# Most Frequen Imputer

df_copy['gender'] = most_fq_imputer.fit_transform(df[['gender']]).ravel()
````

````python
# Missing Indicator

df_copy['annual_income_missing_indicator'] = df_copy['annual_income'].isna().astype(int)
````

````python
# Random Imputer

np.random.seed(42)

missing = df_copy['annual_income'].isna().sum()

if missing > 0:
    random_variable = df_copy['annual_income'].dropna().sample(
        n=missing,
        replace=True,
        random_state=42
    ).values

    df_copy.loc[df_copy['annual_income'].isnull(), 'annual_income'] = random_variable
````

````python
from sklearn.impute import KNNImputer
````

````python
# KNN Imputer

knn = KNNImputer(n_neighbors=2)

df_copy['loan_amount'] = knn.fit_transform(df[['loan_amount']]).ravel()
````

````python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
````

````python
# MICE Imputer

imputer = IterativeImputer(max_iter=10, random_state=42)

df_copy['credit_score'] = imputer.fit_transform(df[['credit_score']]).ravel()
````

## Part D: Outlier Handling

````python
# Z-Score Method

credit_mean = df_copy['credit_score'].mean()
credit_std = df_copy['credit_score'].std()

df_copy['z_score'] = ((df_copy['credit_score'] - credit_mean) / credit_std)

df_copy[['credit_score','z_score']].head()
````

````python
# Remove Outliers

df_clean = df_copy[df_copy['z_score'].abs() <= 3].copy()
````

````python
df_clean.drop(columns='z_score',axis=1,inplace=True)
````

````python
# Percentile Method

lower_bound = df_clean['loan_amount'].quantile(0.05)
upper_bound= df_clean['loan_amount'].quantile(0.95)

lower_bound,upper_bound
````

````python
outlier_mask = (df_clean['loan_amount'] < lower_bound) | (df_clean['loan_amount'] > upper_bound)

print('Outliers Count: ')
df_clean[outlier_mask]['loan_amount'].count()
````

````python
# Remove Ouliers

df_clean = df_clean[~outlier_mask].copy()
````

````python
# Winsorization Method

lower = df_clean['annual_income'].quantile(0.05)
upper= df_clean['annual_income'].quantile(0.95)

df_clean['income_winsorized'] = df_clean['annual_income'].clip(lower=lower,upper=upper)
````

## Part G: Feature Construction & Transformation

````python
from sklearn.preprocessing import FunctionTransformer,PowerTransformer
from sklearn.compose import ColumnTransformer
````

````python
# Functions for ColumnsTransformer

log = ('log_',FunctionTransformer(func=np.log),['spending_ratio'])
reciprocal = ('reciprocal_',FunctionTransformer(func=lambda x: 1 / (x + 1)),['spending_ratio'])
square_root = ('square_root_',FunctionTransformer(func=np.sqrt),['spending_ratio'])
box_cox = ('box-cox',PowerTransformer(method='box-cox'),['loan_amount'])
yeo_johnson = ('yeo-johnson',PowerTransformer(),['annual_income'])

def ratio(x):
    x=x.to_numpy()
    return x[:,[0]]/x[:,[1]]
debt_to_income = ('ratio',FunctionTransformer(func=ratio),['loan_amount','annual_income'])

ct = ColumnTransformer(
    transformers=[
        log,
        reciprocal,
        square_root,
        box_cox,
        yeo_johnson,
        debt_to_income
    ],remainder='drop'
)

columns = [
    'log_spending_ratio',
    'reciprocal_spending_ratio',
    'sqrt_spending_ratio',
    'boxcox_loan_amount',
    'yeojohnson_annual_income',
    'debt_to_income'
]

tf_df = pd.DataFrame(ct.fit_transform(df_clean),columns=columns)
tf_df
````

### 📄 Report Summary

#### 🔹 Missing Value Strategies & Effectiveness
- Numerical features (`age`, `bmi`, `cholesterol`, `glucose`) were handled using **median imputation** to reduce the impact of outliers.  
- Categorical features (`gender`, `region`) were filled using **mode (most frequent) imputation**.  
- This approach preserved data distribution and avoided bias caused by dropping rows.

---

#### 🔹 Outlier Handling Results
- Outliers in `bmi`, `blood_pressure`, `cholesterol`, and `glucose` were detected using the **IQR (Interquartile Range) method**.  
- Extreme values were treated using **capping (winsorization)** to bring them within acceptable limits.  
- This improved data consistency and prevented model distortion.

---

#### 🔹 Encoding Methods Applied
- Categorical variables (`gender`, `region`) were transformed using:
  - **Label Encoding** for binary categories (`gender`)  
  - **One-Hot Encoding** for multi-class categories (`region`)  
- This ensured compatibility with machine learning algorithms.

---

#### 🔹 Scaling / Transformations Applied
- Numerical features were scaled using **Standardization (Z-score scaling)**.  
- This helped normalize feature ranges and improved model performance.  

---

#### 🔹 Newly Engineered Features
- Created **BMI categories** (Underweight, Normal, Overweight, Obese) for better interpretability.  
- Generated **risk-related ratios** (e.g., cholesterol-to-glucose ratio) to capture deeper relationships.  
- These features enhanced predictive power.

---

#### 🔹 Final Dataset Shape & Readiness
- Final dataset contains:
  - Cleaned and imputed values
